# Оценки статистической связи на основе таблицы сопряженности

Для описания статистической зависимости для категориальных признаков или дискретных величин с малым количеством уникальных значений обычно используют частоты совместных распределений. Одним из наиболее важных инструментов является **таблица (матрица) сопряженности**. На основе такой таблицы считаются коэффициенты, позволяющие оценить наличие и силу статистической связи между этими признаками. В литературе такая связь часто называется **ассоциацией**.

**Таблица сопряженности (Contingency Tables)** — это форма представления совместного распределения двух или более категориальных признаков, позволяющая анализировать взаимосвязь между ними. Структурно таблица представляет собой матрицу, в которой строки соответствуют категориям одного признака, а столбцы — категориям другого. Элементы такой матрицы содержат абсолютные или относительные частоты совместного появления соответствующих пар категорий. На основе таблицы сопряженности вычисляются маргинальные распределения, характеризующие частоты признаков в отрыве от друг друга, и условные распределения, что является основой для проверки статистических гипотез о наличии или отсутствии стохастической зависимости между признаками

**Бинарная таблица сопряженности** представляет собой частный случай таблицы сопряженности для анализа взаимосвязи двух дихотомических (бинарных) признаков, каждый из которых имеет только две категории. Структурно она формирует матрицу 2×2, в ячейках которой содержатся частоты совместного появления всех возможных комбинаций категорий.

Например, для признаков $A (A_1, A₂_2)$ и B $(B_1, B_2)$ таблица принимает вид:

|       | $B_1$    | $B_2$    | $\Sigma$     |
|-------|-------|-------|-------|
| __$A_1$__ | $n_{11}$   | $n_{12}$   | $n_{1+}$   |
| __$A_2$__ | $n_{21}$   | $n_{22}$   | $n_{2+}$   |
| __$\Sigma$__  | $n_{+1}$   | $n_{+2}$   | $N$     |

Где:
*   `n₁₁`, `n₁₂`, `n₂₁`, `n₂₂` — наблюдаемые частоты совместного появления соответствующих пар категорий.
*   `n₁₊`, `n₂₊` — маргинальные частоты для признака A.
*   `n₊₁`, `n₊₂` — маргинальные частоты для признака B.
*   `N` — объем выборки.



Или чаще таблицу записывают вот в таком виде:

|       | $B_1$    | $B_2$    |
|-------|-------|-------|
| __$A_1$__ | a     | b     |
| __$A_2$__ | c     | d     |



```python
import numpy as np
import pandas as pd
from scipy.stats.contingency import association
import statsmodels.api as sm

# 1. СОЗДАНИЕ ТАБЛИЦЫ СОПРЯЖЕННОСТИ

# Создаем таблицу 2x2 вручную
data1 = np.array([[50, 450],   # Вакцинированы: заболели, не заболели
                [150, 350]]) # Невакцинированы: заболели, не заболели

def create_contingency_table(data):

    # Пример: Исследование эффективности вакцинации
    # Создаем DataFrame для красивого отображения
    df = pd.DataFrame(data,
                     index=['Вакцинированы', 'Невакцинированы'],
                     columns=['Заболели', 'Не заболели'])

    print(df)
    print(f"Общий объем выборки: {data.sum()}")


create_contingency_table(data1)
```

## Отношение шансов (Odds Ratio)

**Отношение шансов (Odds Ratio, OR)** — фундаментальная мера связи в анализе бинарных данных, широко используемая в эпидемиологии, медицинских исследованиях и статистике.

**Шансы** события $E$ определяются как отношение вероятности его наступления к вероятности его ненаступления:

$$Odds(E) = \frac{P(E)}{1 - P(E)}$$

Предположим, что есть таблица сопряженности полученная при исследовании эффективности вакцинации против оспы:

|               | Заболели ($D^+$) | Здоровы ($D^-$) |
|---------------|------------------|-----------------|
| **Вакцинированы ($V^+$)** | $a$              | $b$             |
| **Невакцинированы ($V^-$)** | $c$              | $d$             |

Шансы заболевания среди вакцинированных:
$$Odds_1 = \frac{P(D^+|V^+)}{P(D^-|V^+)} = \frac{a/(a+b)}{b/(a+b)} = \frac{a}{b}$$

Шансы заболевания среди невакцинированных:
$$Odds_2 = \frac{P(D^+|V^-)}{P(D^-|V^-)} = \frac{c/(c+d)}{d/(c+d)} = \frac{c}{d}$$

Отношение шансов:
$$OR = \frac{Odds_1}{Odds_2} = \frac{a/b}{c/d} = \frac{a \cdot d}{b \cdot c}$$


Рассмотрим численный пример

|               | Заболели оспой | Не заболели |
|---------------|----------------|-------------|
| **Вакцинированы** | 50             | 450         |
| **Невакцинированы** | 150            | 350         |



$$OR = \frac{50 \cdot 350}{450 \cdot 150} = \frac{17500}{67500} \approx 0.259$$


Отношение шансов $OR \approx 0.259$ означает, что шансы заболеть оспой среди вакцинированных составляют примерно 25.9% от шансов заболеть среди невакцинированных. Иными словами, вакцинация снижает шансы заболевания примерно в 3.86 раза ($1/0.259 \approx 3.86$).


- **$OR = 1$**: отсутствие связи между переменными
- **$OR > 1$**: положительная ассоциация (фактор увеличивает шансы события)
- **$OR < 1$**: отрицательная ассоциация (фактор уменьшает шансы события)


- **$OR = 2.0$**: шансы события в экспонированной группе в 2 раза выше
- **$OR = 0.5$**: шансы события в экспонированной группе в 2 раза ниже
- **$OR = 3.0$**: шансы события в экспонированной группе в 3 раза выше

**Важное замечание**: При интерпретации OR важно учитывать базовую вероятность события, так как одинаковые значения OR могут соответствовать разным абсолютным эффектам в зависимости от распространенности исхода.

```python
# 2. ОТНОШЕНИЕ ШАНСОВ (ODDS RATIO)

def calculate_odds_ratio(table):

    # Извлекаем значения из таблицы
    a, b, c, d = table[0,0], table[0,1], table[1,0], table[1,1]

    # Шансы заболевания среди вакцинированных
    odds_vaccinated = a / b

    # Шансы заболевания среди невакцинированных
    odds_unvaccinated = c / d

    # Отношение шансов
    odds_ratio = (a * d) / (b * c)

    print("ОТНОШЕНИЕ ШАНСОВ (ODDS RATIO):")
    print(f"Шансы заболевания среди вакцинированных: {odds_vaccinated:.3f}")
    print(f"Шансы заболевания среди невакцинированных: {odds_unvaccinated:.3f}")
    print(f"Отношение шансов (OR): {odds_ratio:.3f}")

    # Интерпретация
    if odds_ratio > 1:
        print(f"Вакцинация УВЕЛИЧИВАЕТ шансы заболевания в {odds_ratio:.1f} раз")
    elif odds_ratio < 1:
        print(f"Вакцинация УМЕНЬШАЕТ шансы заболевания в {1/odds_ratio:.1f} раз")
    else:
        print("Вакцинация НЕ ВЛИЯЕТ на шансы заболевания")

    return odds_ratio

calculate_odds_ratio(data1)
```

## Характеристики статистической связи на основе таблиц сопряженности

**Коэффициенты ассоциации Юла Q и Y** применяются для анализа связи в бинарных таблицах сопряженности.

**Yule's Q** (коэффициент ассоциации Юла) вычисляется по формуле:
$$Q = \frac{ad - bc}{ad + bc}$$

**Yule's Y** (коэффициент коинтеграции Юла) определяется как:
$$Y = \frac{\sqrt{ad} - \sqrt{bc}}{\sqrt{ad} + \sqrt{bc}}$$


Оба коэффициента измеряют силу и направление ассоциации между дихотомическими признаками. Значения изменяются в диапазоне от -1 до +1, где:
- +1 указывает на полную положительную ассоциацию
- -1 указывает на полную отрицательную ассоциацию  
- 0 свидетельствует об отсутствии ассоциации

1. Коэффициент Q более чувствителен к крайним значениям и может достигать границ ±1 при наличии нулевых ячеек
2. Коэффициент Y менее чувствителен к крайним значениям и обеспечивает более консервативную оценку
3. Оба коэффициента симметричны и инвариантны относительно умножения на константу
4. Интерпретация силы связи: |0-0,3| - слабая, |0,3-0,7| - умеренная, |0,7-1| - сильная


```python
# 3. КОЭФФИЦИЕНТЫ ЮЛА Q и Y

def calculate_yule_coefficients(table):

    a, b, c, d = table[0,0], table[0,1], table[1,0], table[1,1]

    # Коэффициент Юла Q
    yule_q = (a*d - b*c) / (a*d + b*c)

    # Коэффициент Юла Y
    yule_y = (np.sqrt(a*d) - np.sqrt(b*c)) / (np.sqrt(a*d) + np.sqrt(b*c))

    print("КОЭФФИЦИЕНТЫ АССОЦИАЦИИ ЮЛА:")

    print(f"Коэффициент Юла Q: {yule_q:.3f}")
    print(f"Коэффициент Юла Y: {yule_y:.3f}")

    # Интерпретация силы связи
    def interpret_strength(coef):
        abs_coef = abs(coef)
        if abs_coef < 0.3:
            return "слабая"
        elif abs_coef < 0.7:
            return "умеренная"
        else:
            return "сильная"

    print(f"Сила связи (Q): {interpret_strength(yule_q)}")
    print(f"Сила связи (Y): {interpret_strength(yule_y)}")

    return yule_q, yule_y

calculate_yule_coefficients(data1)
```

**Коэффициент $\phi$  Пирсона** является мерой связи для бинарных таблиц сопряженности 2×2. Пусть таблица имеет вид:

Коэффициент $\phi$ вычисляется как:
$$\phi = \frac{ad - bc}{\sqrt{(a+b)(c+d)(a+c)(b+d)}}$$

Коэффициент принимает значения в диапазоне от -1 до +1. При этом значения трактуются следующим образом:
   - 0: отсутствие линейной связи между признаками
   - +1: полная положительная ассоциация
   - -1: полная отрицательная ассоциация

Знак коэффициента указывает на направление ассоциации между категориями

- Коэффициент $\phi$ симметричен — не зависит от того, какой признак рассматривается как независимый
- По абсолютной величине интерпретируется аналогично коэффициенту корреляции: |0-0,3| — слабая связь, |0,3-0,7| — умеренная, |0,7-1| — сильная
- Особенно полезен при сравнении силы связи в таблицах одинакового размера, но с разным объемом выборки
- Коэффициент $\phi$ непосредственно связан с критерием хи-квадрат Пирсона (об этом см. подробнее раздел "Тесты на значимость")


```python
# 4. КОЭФФИЦИЕНТ φ ПИРСОНА

def calculate_phi_coefficient(table):

    a, b, c, d = table[0,0], table[0,1], table[1,0], table[1,1]

    # Формула коэффициента φ
    numerator = a*d - b*c
    denominator = np.sqrt((a+b)*(c+d)*(a+c)*(b+d))
    phi = numerator / denominator

    print("КОЭФФИЦИЕНТ φ ПИРСОНА:")

    print(f"Коэффициент φ: {phi:.3f}")

    # Интерпретация
    if phi > 0:
        direction = "положительная"
    elif phi < 0:
        direction = "отрицательная"
    else:
        direction = "отсутствует"

    abs_phi = abs(phi)
    if abs_phi < 0.3:
        strength = "слабая"
    elif abs_phi < 0.7:
        strength = "умеренная"
    else:
        strength = "сильная"

    print(f"Направление связи: {direction}")
    print(f"Сила связи: {strength}")

    return phi

calculate_phi_coefficient(data1)
```

Эти показатели являются достаточно распространенными и простыми коэффициентами, которые присутствуют в таких средах как R и Matlab. Но, что удивительно, в популярных статистических библиотеках Python их нет.  

Однако в библиотеке *scipy.stats* есть метод расчета коэффициента сопряженности Пирсона, который по модулю совпадает с коэффициентом $\phi$ Пирсона (хотя чаще используется для мультиклассовые таблиц сопряженности) но всегда принимает положительные значения.

**Мультиклассовые таблицы сопряженности** представляют собой любые таблицы размером $r 𝙭 c$, т.е. где число уникальных значений которые могут принять величины $X$ и $Y$ соответсвенно равно $r$ и $c$.

```python
# 5. МУЛЬТИКЛАССОВЫЕ ТАБЛИЦЫ

 # Пример: Уровень образования vs Уровень дохода

data2 = np.array([[20, 35, 15],   # Среднее образование: низкий, средний, высокий доход
                     [15, 40, 25],   # Высшее образование
                     [5,  20, 30]])  # Аспирантура


def create_multiclass_table(data):


    df = pd.DataFrame(data,
                     index=['Среднее', 'Высшее', 'Аспирантура'],
                     columns=['Низкий доход', 'Средний доход', 'Высокий доход'])

    print("МУЛЬТИКЛАССОВАЯ ТАБЛИЦА СОПРЯЖЕННОСТИ 3x3:")

    print(df)
    print(f"Общий объем выборки: {data.sum()}")

    return data

create_multiclass_table(data2)
```

## Коэффициент сопряженности Пирсона

**Коэффициент сопряженности Пирсона (Pearson's Contingency Coefficient)** — это мера силы связи между двумя категориальными переменными в таблице сопряженности произвольного размера $r × c$.

Для таблицы сопряженности размерности $r × c$ коэффициент сопряженности Пирсона вычисляется по формуле:

$$C = \sqrt{\frac{\chi^2}{\chi^2 + n}}$$

где:
- $\chi^2$ — значение статистики хи-квадрат Пирсона
- $n$ — общий объем выборки
- $r$ — количество строк
- $c$ — количество столбцов

Расчет хи-квадрат Пирсона:

$$\chi^2 = \sum_{i=1}^{r} \sum_{j=1}^{c} \frac{(O_{ij} - E_{ij})^2}{E_{ij}}$$

где:
- $O_{ij}$ — наблюдаемая частота в ячейке $(i,j)$
- $E_{ij}$ — ожидаемая частота в ячейке $(i,j)$, вычисляемая как:
  $$E_{ij} = \frac{(сумма\ строки\ i) × (сумма\ столбца\ j)}{n}$$


Диапазон значений:
- **$C = 0$**: полная независимость переменных
- **$C > 0$**: наличие связи между переменными
- **0.0-0.1**: очень слабая связь
- **0.1-0.3**: слабая связь  
- **0.3-0.5**: умеренная связь
- **0.5-0.7**: сильная связь
- **>0.7**: очень сильная связь (относительно максимального значения)
- **Максимальное значение**: зависит от размера таблицы

Максимально возможное значение:
$$C_{max} = \sqrt{\frac{m-1}{m}}$$
где $m = \min(r, c)$ — минимальная размерность таблицы.

Примеры максимальных значений:
- Для таблицы $2 × 2$: $C_{max} = \sqrt{\frac{1}{2}} ≈ 0.707$
- Для таблицы $3 × 3$: $C_{max} = \sqrt{\frac{2}{3}} ≈ 0.816$
- Для таблицы $4 × 4$: $C_{max} = \sqrt{\frac{3}{4}} = 0.866$


**Следует учесть**, что коэффициент сопряженности Пирсона зависит от размера таблицы, что затрудняет анлиз значений полученных по признакам с разным количестов интервалов или уникальных значений, и принимает всегда положительные значения, что не позволяет оценить направление статистической связи.

**Рекомендация**: При работе с таблицами разного размера предпочтительнее использовать коэффициент Крамера V, так как он позволяет проводить сравнения между исследованиями с разной структурой данных.

```python
# 6. КОЭФФИЦИЕНТ СОПРЯЖЕННОСТИ ПИРСОНА

def calculate_pearson_contingency(table):

    # Используем встроенную функцию из scipy
    pearson_coef = association(table, method="pearson")

    print("КОЭФФИЦИЕНТ СОПРЯЖЕННОСТИ ПИРСОНА:")

    print(f"Коэффициент: {pearson_coef:.3f}")

    # Интерпретация
    if pearson_coef < 0.1:
        strength = "очень слабая"
    elif pearson_coef < 0.3:
        strength = "слабая"
    elif pearson_coef < 0.5:
        strength = "умеренная"
    elif pearson_coef < 0.7:
        strength = "сильная"
    else:
        strength = "очень сильная"

    print(f"Сила связи: {strength}")

    # Максимально возможное значение для данной таблицы
    r, c = table.shape
    m = min(r, c)
    max_possible = np.sqrt((m-1)/m)
    print(f"Максимально возможное значение для таблицы {r}x{c}: {max_possible:.3f}")

    return pearson_coef

calculate_pearson_contingency(data2)
```

## Коэффициент сопряжнности Чупрова


**Коэффициент сопряженности Чупрова (Tschuprow's T)** представляет собой симметричную меру связи между двумя категориальными переменными в таблице сопряженности размерности $r × c$. Данный коэффициент был разработан Александром Чупровым как модификация коэффициента фи Пирсона для таблиц произвольного размера, устраняющая зависимость от размерности таблицы.

Математически коэффициент Чупрова определяется по формуле:

$$T = \sqrt{\frac{\chi^2}{n \cdot \sqrt{(r-1)(c-1)}}}$$

где $\chi^2$ — значение статистики хи-квадрат Пирсона, $n$ — объем выборки, $r$ и $c$ — количество строк и столбцов соответственно. Расчет хи-квадрат осуществляется стандартным образом:

$$\chi^2 = \sum_{i=1}^{r} \sum_{j=1}^{c} \frac{(O_{ij} - E_{ij})^2}{E_{ij}}$$

где $O_{ij}$ — наблюдаемая частота, а $E_{ij} = \frac{(сумма\_строки\_i) × (сумма\_столбца\_j)}{n}$ — ожидаемая частота.

Коэффициент Чупрова обладает важным свойством симметричности: $T(X,Y) = T(Y,X)$, что делает его универсальным для анализа взаимосвязи между переменными без выделения зависимой и независимой переменной. Диапазон значений коэффициента ограничен интервалом от 0 до 1, где $T = 0$ свидетельствует о полной независимости переменных, а значения, приближающиеся к 1, указывают на сильную статистическую связь. В отличие от коэффициента сопряженности Пирсона, максимальное значение коэффициента Чупрова всегда равно 1 независимо от размерности таблицы, что позволяет сравнивать силу связи между таблицами разного размера.

Интерпретация коэффициента осуществляется по качественной шкале:

- значения $0 < T < 0.1$ указывают на очень слабую связь,
- $0.1 ≤ T < 0.3$ — на слабую связь,
- $0.3 ≤ T < 0.5$ — на умеренную связь,
- $0.5 ≤ T < 0.7$ — на заметную связь,
- $0.7 ≤ T < 0.9$ — на сильную связь, и
- $T ≥ 0.9$ — на очень сильную связь.

Следует отметить, что коэффициент Чупрова, как и другие меры на основе хи-квадрат, не указывает направление связи и чувствителен к объему выборки.

По сравнению с коэффициентом Крамера $V = \sqrt{\frac{\chi^2}{n \cdot min(r-1, c-1)}}$, коэффициент Чупрова является более консервативной мерой, поскольку в знаменателе используется среднее геометрическое $\sqrt{(r-1)(c-1)}$ вместо минимального значения $(r-1)$ или $(c-1)$. Это делает коэффициент Чупрова особенно полезным при работе с таблицами, имеющими существенно различное количество категорий по строкам и столбцам, где он обеспечивает более сбалансированную оценку силы связи.

```python
# 7. КОЭФФИЦИЕНТ ЧУПРОВА

def calculate_tschuprow_t(table):

    # Используем встроенную функцию из scipy
    tschuprow_coef = association(table, method="tschuprow")

    print("КОЭФФИЦИЕНТ СОПРЯЖЕННОСТИ ЧУПРОВА:")

    print(f"Коэффициент T: {tschuprow_coef:.3f}")

    # Интерпретация (аналогична другим коэффициентам)
    abs_coef = abs(tschuprow_coef)
    if abs_coef < 0.1:
        strength = "очень слабая"
    elif abs_coef < 0.3:
        strength = "слабая"
    elif abs_coef < 0.5:
        strength = "умеренная"
    elif abs_coef < 0.7:
        strength = "заметная"
    elif abs_coef < 0.9:
        strength = "сильная"
    else:
        strength = "очень сильная"

    print(f"Сила связи: {strength}")

    return tschuprow_coef

calculate_tschuprow_t(data2)
```

## V Крамера

**Коэффициент V Крамера (Cramér's V)** является симметричной мерой силы связи между двумя номинальными переменными в таблице сопряженности произвольной размерности $r × c$. Разработанный шведским математиком Харальдом Крамером на основе статистики хи-квадрат Пирсона, данный коэффициент устраняет зависимость от объема выборки и размерности таблицы, что делает его особенно ценным для сравнительного анализа таблиц различной структуры.

Математически коэффициент Крамера определяется как нормализованная версия статистики хи-квадрат:
$$V = \sqrt{\frac{\chi^2}{n \cdot (m - 1)}}$$

где $\chi^2$ — значение статистики хи-квадрат Пирсона, вычисляемое по формуле:

$$\chi^2 = \sum_{i=1}^{r} \sum_{j=1}^{c} \frac{(O_{ij} - E_{ij})^2}{E_{ij}}$$

где $n$ представляет объем выборки, а $m = \min(r, c)$ обозначает минимальную размерность таблицы (меньшее из количества строк или столбцов). В знаменателе множитель $(m - 1)$ обеспечивает нормализацию коэффициента в диапазон от 0 до 1.

Интерпретация коэффициента Крамера осуществляется по качественной шкале, где $V = 0$ свидетельствует о полном отсутствии связи между переменными, а значение $V = 1$ указывает на perfect association — идеальную предсказуемость одной переменной по другой. Практическая интерпретация силы связи следует общепринятым стандартам:

- значения $0 < V < 0.1$ рассматриваются как очень слабая связь,
- $0.1 ≤ V < 0.3$ — слабая связь,
- $0.3 ≤ V < 0.5$ — умеренная связь,
- $0.5 ≤ V < 0.7$ — заметная связь,
- $0.7 ≤ V < 0.9$ — сильная связь,
- $V ≥ 0.9$ — очень сильная связь.

В случае таблицы 2×2 коэффициент Крамера эквивалентен коэффициенту фи Пирсона по абсолютному значению: $V = |\phi|$.

Важным преимуществом коэффициента Крамера является его инвариантность относительно объемов выборки и симметричность — $V(X,Y) = V(Y,X)$, что позволяет объективно сравнивать силу связи в исследованиях с различным дизайном и объемом данных. По сравнению с коэффициентом сопряженности Пирсона, коэффициент Крамера не зависит от размерности таблицы, а в сравнении с коэффициентом Чупрова он обеспечивает более консервативную оценку для таблиц с неравным количеством строк и столбцов. Широкое применение коэффициент находит в социологии, маркетинговых исследованиях, медицине и анализе качества классификационных моделей, где требуется стандартизированная оценка силы взаимосвязи категориальных переменных.

```python
# 8. КОЭФФИЦИЕНТ V КРАМЕРА

def calculate_cramers_v(table):

    cramers_v = association(table, method="cramer")

    print("КОЭФФИЦИЕНТ V КРАМЕРА:")

    print(f"Коэффициент V: {cramers_v:.3f}")

    # Интерпретация
    if cramers_v < 0.1:
        strength = "очень слабая"
    elif cramers_v < 0.3:
        strength = "слабая"
    elif cramers_v < 0.5:
        strength = "умеренная"
    elif cramers_v < 0.7:
        strength = "заметная"
    elif cramers_v < 0.9:
        strength = "сильная"
    else:
        strength = "очень сильная"

    print(f"Сила связи: {strength}")

    return cramers_v

calculate_cramers_v(data2)
```
